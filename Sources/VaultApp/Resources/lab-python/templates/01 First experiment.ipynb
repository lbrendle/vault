# Make a question executable

A notebook is a place to think, test, and keep the evidence. Every cell here runs on this device. Start with **Run all**, then change the experiment.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
x = np.linspace(0, 4 * np.pi, 160)
signal = np.sin(x)
observed = signal + rng.normal(0, 0.24, len(x))
print(f"{len(x)} observations · generated locally")
```

## Look before you model

How much of the variation comes from the underlying signal, and how much comes from noise? Change `0.24` above and run the cells again.

```python
fig, ax = plt.subplots(figsize=(9, 3.5))
ax.scatter(x, observed, s=16, alpha=0.48, color="#759b93", label="Observed")
ax.plot(x, signal, color="#344f62", lw=2.4, label="Known signal")
ax.set(xlabel="Time", ylabel="Amplitude")
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, loc="upper right")
fig.tight_layout()
plt.show()
```
