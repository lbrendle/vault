Use symbolic derivations alongside numerical checks. These libraries are installed on this device.

```python
import sympy as sp
x = sp.symbols("x")
f = sp.sin(x) * sp.exp(-x / 3)
sp.diff(f, x)
```

```python
import numpy as np
a = np.array([[3., 1.], [1., 2.]])
b = np.array([9., 8.])
solution = np.linalg.solve(a, b)
np.testing.assert_allclose(a @ solution, b)
solution
```
